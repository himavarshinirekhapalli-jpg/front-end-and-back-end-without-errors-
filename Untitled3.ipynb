```python
import numpy as np
import matplotlib.pyplot as plt
import ipywidgets as widgets
from IPython.display import display, clear_output
```

```python
def rotation_matrix(angle):
    theta = np.radians(angle)
    return np.array([
        [np.cos(theta), -np.sin(theta)],
        [np.sin(theta),  np.cos(theta)]
    ])

def rotate_point(x, y, angle):
    R = rotation_matrix(angle)
    point = np.array([x, y])
    return R @ point

def check_orthogonal(R):
    return np.allclose(R.T @ R, np.eye(2))

def calculate_properties(R):
    return {
        "RᵀR": R.T @ R,
        "Determinant": np.linalg.det(R),
        "Orthogonal": check_orthogonal(R)
    }
```

```python
def plot_rotation(x, y, angle):
    original = np.array([x, y])
    rotated = rotate_point(x, y, angle)

    plt.figure(figsize=(7, 7))

    plt.axhline(0, linewidth=1)
    plt.axvline(0, linewidth=1)

    plt.scatter(original[0], original[1], s=100, label="Original Point")
    plt.scatter(rotated[0], rotated[1], s=100, label="Rotated Point")

    plt.plot(
        [0, original[0]],
        [0, original[1]],
        linestyle="--"
    )

    plt.plot(
        [0, rotated[0]],
        [0, rotated[1]],
        linestyle="--"
    )

    plt.annotate(
        f"P({original[0]:.2f}, {original[1]:.2f})",
        original
    )

    plt.annotate(
        f"P'({rotated[0]:.2f}, {rotated[1]:.2f})",
        rotated
    )

    limit = max(abs(x), abs(y), abs(rotated[0]), abs(rotated[1]), 2) + 2

    plt.xlim(-limit, limit)
    plt.ylim(-limit, limit)
    plt.xlabel("X-axis")
    plt.ylabel("Y-axis")
    plt.title(f"Screen Rotation: {angle}°")
    plt.grid(True)
    plt.legend()
    plt.axis("equal")
    plt.show()
```

```python
print("=" * 60)
print("     SCREEN ROTATION AS ORTHOGONAL TRANSFORMATION")
print("=" * 60)
print()
print("Mathematical model of phone screen rotation")
print("Rotation preserves distance, angles and shape.")
```

```python
x_input = widgets.FloatText(
    value=2,
    description="X:",
    style={"description_width": "50px"}
)

y_input = widgets.FloatText(
    value=1,
    description="Y:",
    style={"description_width": "50px"}
)

angle_input = widgets.Dropdown(
    options=[
        ("90° Clockwise", -90),
        ("90° Counterclockwise", 90),
        ("180°", 180),
        ("270° Counterclockwise", 270)
    ],
    value=90,
    description="Rotation:",
    style={"description_width": "70px"}
)

run_button = widgets.Button(
    description="Rotate Screen",
    button_style="primary"
)

output = widgets.Output()

display(
    widgets.VBox([
        x_input,
        y_input,
        angle_input,
        run_button,
        output
    ])
)
```

```python
def run_project(button):
    with output:
        clear_output()

        x = x_input.value
        y = y_input.value
        angle = angle_input.value

        R = rotation_matrix(angle)
        rotated = rotate_point(x, y, angle)
        properties = calculate_properties(R)

        print("ROTATION MATRIX")
        print("----------------")
        print(np.round(R, 4))
        print()

        print("ORIGINAL POINT")
        print("----------------")
        print(f"P = ({x:.4f}, {y:.4f})")
        print()

        print("ROTATED POINT")
        print("----------------")
        print(f"P' = ({rotated[0]:.4f}, {rotated[1]:.4f})")
        print()

        print("ORTHOGONALITY CHECK")
        print("----------------")
        print("RᵀR =")
        print(np.round(properties["RᵀR"], 4))
        print()

        print(f"Determinant = {properties['Determinant']:.4f}")
        print(f"Orthogonal Matrix = {properties['Orthogonal']}")
        print()

        original_distance = np.linalg.norm([x, y])
        rotated_distance = np.linalg.norm(rotated)

        print("DISTANCE PRESERVATION")
        print("----------------")
        print(f"Original distance = {original_distance:.4f}")
        print(f"Rotated distance  = {rotated_distance:.4f}")
        print(
            "Distance preserved =",
            np.isclose(original_distance, rotated_distance)
        )

        plot_rotation(x, y, angle)

run_button.on_click(run_project)
```

```python
print("PORTRAIT → LANDSCAPE EXAMPLE")
print("=" * 40)

R = rotation_matrix(90)

print("Rotation Matrix:")
print(np.round(R, 2))

point = np.array([2, 1])
new_point = R @ point

print()
print(f"Original point : {point}")
print(f"Rotated point  : {np.round(new_point, 2)}")

print()
print("RᵀR =")
print(np.round(R.T @ R, 2))

print()
print("det(R) =", round(np.linalg.det(R), 2))
print("Orthogonal =", check_orthogonal(R))
```
